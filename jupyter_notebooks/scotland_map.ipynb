```python
# Download and load Scotland unemployment table
import json
import requests
import pandas as pd
import altair as alt
import os

excel_url = 'https://www.gov.scot/binaries/content/documents/govscot/publications/statistics/2022/03/sub-scotland-economic-statistics-database/documents/sub-scotland-economic-statistics-database/sub-scotland-economic-statistics-database/govscot%3Adocument/Sub-Scotland%2BEconomic%2BStatistics%2BDatabase.xlsx'
excel_path = '../data/sub_scotland_economic_stats.xlsx'

os.makedirs('../data', exist_ok=True)
response = requests.get(excel_url, timeout=60)
if response.status_code != 200:
    raise RuntimeError(f'Failed to download Excel file (status {response.status_code})')
with open(excel_path, 'wb') as f:
    f.write(response.content)

df = pd.read_excel(
    excel_path,
    sheet_name='Table 16',
    usecols='C:X',
    skiprows=11,
    nrows=11
)

df.head()
```

```python
# Inspect structure of regions and year columns
region_col = df.columns[0]
year_cols = df.columns[1:].tolist()

summary_cols = pd.DataFrame(
    {
        'region_column': [region_col],
        'year_columns_count': [len(year_cols)]
    }
)

year_cols_preview = pd.DataFrame({'year_columns': year_cols}).head(10)

summary_cols
year_cols_preview
```

```python
# Reshape unemployment data to tidy long format and filter relevant regions/years
region_col = df.columns[0]
df_clean = df.rename(columns={region_col: 'region'})

unemployment_long = df_clean.melt(
    id_vars=['region'],
    var_name='year',
    value_name='unemployment_rate'
)

unemployment_long['year'] = pd.to_numeric(unemployment_long['year'], errors='coerce')
unemployment_long['unemployment_rate'] = pd.to_numeric(unemployment_long['unemployment_rate'], errors='coerce')
unemployment_long = unemployment_long.dropna()
unemployment_long['region'] = unemployment_long['region'].str.strip()

exclude_regions = ['Scotland', 'UK', 'Islands & Remote Rural']
unemployment_long = unemployment_long[~unemployment_long['region'].isin(exclude_regions)]

all_years = sorted(unemployment_long['year'].unique())
if len(all_years) > 4:
    last_4_years = all_years[-4:]
    unemployment_long = unemployment_long[unemployment_long['year'].isin(last_4_years)]

unemployment_long.head(15)
```

```python
# Build GeoJSON for Regional Economic Partnerships from council geometries
from shapely.geometry import shape, mapping
from shapely.ops import unary_union

graphs_dir = '../data'
geojson_url = 'https://raw.githubusercontent.com/martinjc/UK-GeoJSON/master/json/administrative/sco/lad.json'
response = requests.get(geojson_url, timeout=30)
if response.status_code != 200:
    raise RuntimeError(f'Failed to fetch council GeoJSON (status {response.status_code})')

council_geo = response.json()
rep_mapping = {
    'Aberdeen City and Aberdeenshire': ['Aberdeen City', 'Aberdeenshire'],
    'Ayrshire': ['East Ayrshire', 'North Ayrshire', 'South Ayrshire'],
    'Borderlands': ['Dumfries and Galloway', 'Scottish Borders'],
    'Edinburgh and South East Scotland': ['City of Edinburgh', 'East Lothian', 'Midlothian', 'West Lothian', 'Scottish Borders', 'Fife'],
    'Glasgow City Region': ['Glasgow City', 'East Dunbartonshire', 'East Renfrewshire', 'Inverclyde', 'North Lanarkshire', 'Renfrewshire', 'South Lanarkshire', 'West Dunbartonshire'],
    'Inverness and Highland': ['Highland', 'Moray', 'Argyll and Bute', 'Na h-Eileanan Siar'],
    'Stirling and Clackmannanshire': ['Stirling', 'Clackmannanshire', 'Falkirk'],
    'Tay Cities': ['Dundee City', 'Angus', 'Perth and Kinross', 'Fife']
}

council_shapes = {feature['properties']['LAD13NM']: shape(feature['geometry']) for feature in council_geo['features']}
rep_features = []
rep_counts = []

for rep_name, councils in rep_mapping.items():
    matching_shapes = []
    for council in councils:
        if council in council_shapes:
            matching_shapes.append(council_shapes[council])
        else:
            for available_council in council_shapes.keys():
                if council.lower() in available_council.lower() or available_council.lower() in council.lower():
                    matching_shapes.append(council_shapes[available_council])
                    break
    if matching_shapes:
        merged_geometry = unary_union(matching_shapes)
        rep_features.append({
            'type': 'Feature',
            'properties': {'name': rep_name},
            'geometry': mapping(merged_geometry)
        })
        rep_counts.append({'rep_name': rep_name, 'councils_merged': len(matching_shapes)})

scotland_rep_geo = {'type': 'FeatureCollection', 'features': rep_features}
rep_geojson_path = '../data/scotland_regional_partnerships.geojson'
with open(rep_geojson_path, 'w', encoding='utf-8') as f:
    json.dump(scotland_rep_geo, f, ensure_ascii=False, indent=2)

pd.DataFrame(rep_counts).head()
```

```python
# Create choropleth map with year selector and save specification
if len(unemployment_long) == 0:
    raise ValueError('No unemployment data available for visualization')


def normalize_region_name(name):
    name = str(name).lower().strip()
    name_mapping = {
        'aberdeen city region': 'aberdeen city and aberdeenshire',
        'ayrshires': 'ayrshire',
        'south of scotland': 'borderlands',
        'edinburgh and south east scotland city region': 'edinburgh and south east scotland',
        'glasgow city region': 'glasgow city region',
        'highlands and islands': 'inverness and highland',
        'forth valley region': 'stirling and clackmannanshire',
        'tay cities region': 'tay cities'
    }
    for data_name, geo_name in name_mapping.items():
        if data_name in name:
            return geo_name
    return name

years = sorted(unemployment_long['year'].unique())
if len(years) == 0:
    raise ValueError('No years found in the data')

default_year = 2019 if 2019 in years else years[-1]
year_slider = alt.binding_range(min=int(years[0]), max=int(years[-1]), step=1, name='Year')
year_select = alt.param(name='selected_year', bind=year_slider, value=int(default_year))

combined_data = []
for feature in scotland_rep_geo['features']:
    geo_name = feature['properties']['name']
    geo_name_norm = normalize_region_name(geo_name)
    for _, row in unemployment_long.iterrows():
        data_name_norm = normalize_region_name(row['region'])
        if geo_name_norm == data_name_norm:
            combined_data.append({
                'type': feature['type'],
                'geometry': feature['geometry'],
                'properties': feature['properties'],
                'region': geo_name,
                'year': int(row['year']),
                'unemployment_rate': float(row['unemployment_rate']) * 100
            })

if len(combined_data) == 0:
    raise ValueError('No matching data found between GeoJSON and unemployment data')

all_rates = [d['unemployment_rate'] for d in combined_data]
min_rate, max_rate = min(all_rates), max(all_rates)

title_block = {
    'text': 'Scotland Unemployment Rate By Region (Before & During COVID-19)',
    'subtitle': [
        'Source: Scottish Government (https://www.gov.scot/publications/sub-scotland-economic-statistics-database/)',
        'Map data: https://raw.githubusercontent.com/martinjc/UK-GeoJSON/master/json/administrative/sco/lad.json'
    ],
    'anchor': 'start'
}

unemployment_map = alt.Chart(alt.Data(values=combined_data)).mark_geoshape(
    stroke='white',
    strokeWidth=1.5
).encode(
    color=alt.Color('unemployment_rate:Q',
                    scale=alt.Scale(scheme='oranges', domain=[min_rate, max_rate]),
                    title='Unemployment Rate (%)',
                    legend=alt.Legend(orient='right', format='.1f')),
    tooltip=[
        alt.Tooltip('region:N', title='Region'),
        alt.Tooltip('unemployment_rate:Q', title='Unemployment Rate (%)', format='.1f'),
        alt.Tooltip('year:N', title='Year')
    ]
).add_params(
    year_select
).transform_filter(
    alt.datum.year == year_select
).properties(
    width=700,
    height=900,
    title=title_block
).project(
    type='mercator'
)

output_json_path = '../graphs/scotland_unemployment_map.json'
os.makedirs('../graphs', exist_ok=True)
unemployment_map.save(output_json_path)

pd.DataFrame({'saved_chart': [output_json_path], 'records': [len(combined_data)]})
unemployment_map
```
